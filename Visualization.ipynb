```python
import pandas as pd

df = pd.read_csv("Students Performance Dataset.csv")
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(x='Gender', y='Final_Score', data=df)
plt.title("Gender vs Final Score")
plt.show()
```

```python
sns.scatterplot(x='Study_Hours_per_Week', y='Final_Score', data=df)
plt.title("Study Hours vs Final Score")
plt.show()
```

```python
sns.histplot(df['Midterm_Score'], kde=True)
plt.title("Midterm Score Distribution")
plt.show()
```

```python
sns.boxplot(x='Sleep_Hours_per_Night', y='Final_Score', data=df)
plt.title("Sleep vs Performance")
plt.show()
```
