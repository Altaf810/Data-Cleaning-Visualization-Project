```python
import pandas as pd

df = pd.read_csv("Students Performance Dataset.csv")
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14,10))

# 1) Gender vs Final Score
plt.subplot(2,2,1)
sns.barplot(x='Gender', y='Final_Score', data=df)
plt.title("Gender vs Final Score")

# 2️) Study Hours vs Final Score
plt.subplot(2,2,2)
sns.scatterplot(x='Study_Hours_per_Week', y='Final_Score', data=df)
plt.title("Study Hours vs Score")

# 3️) Sleep vs Final Score
plt.subplot(2,2,3)
sns.boxplot(x='Sleep_Hours_per_Night', y='Final_Score', data=df)
plt.title("Sleep vs Score")

plt.tight_layout()
plt.show()
```

Dashboard Insights:

1. Gender does not significantly affect final scores.
2. Students who study more tend to score higher.
3. Proper sleep is associated with better performance.
